# %% [markdown]
# # Fraud Detection - Exploratory Data Analysis
# 
# This notebook explores the credit card fraud dataset

# %%
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler

# %%
# Load data
df = pd.read_csv('../data/raw/creditcard.csv')

# %%
# Basic info
df.info()

# %%
# Class distribution
class_dist = df['Class'].value_counts()
print(class_dist)

plt.figure(figsize=(8,6))
sns.barplot(x=class_dist.index, y=class_dist.values)
plt.title('Class Distribution')
plt.xlabel('Class')
plt.ylabel('Count')
plt.show()

# %%
# Transaction amount analysis
plt.figure(figsize=(12,6))
sns.histplot(df[df['Class']==0]['Amount'], bins=50, color='blue', label='Non-Fraud')
sns.histplot(df[df['Class']==1]['Amount'], bins=50, color='red', label='Fraud')
plt.yscale('log')
plt.legend()
plt.title('Transaction Amount Distribution by Class')
plt.show()

# %%
# Time analysis
df['hour'] = df['Time'] % (24*3600) // 3600
df['day'] = (df['Time'] // (24*3600)) % 7

plt.figure(figsize=(12,6))
sns.countplot(x='hour', hue='Class', data=df)
plt.title('Transactions by Hour of Day')
plt.show()

# %%
# Correlation matrix
plt.figure(figsize=(16,12))
sns.heatmap(df.corr(), cmap='coolwarm', center=0)
plt.title('Correlation Matrix')
plt.show()

# %%
# PCA components analysis
pca_cols = [f'V{i}' for i in range(1,29)]
plt.figure(figsize=(16,20))
for i, col in enumerate(pca_cols):
    plt.subplot(7,4,i+1)
    sns.boxplot(x='Class', y=col, data=df)
plt.tight_layout()
plt.show()
